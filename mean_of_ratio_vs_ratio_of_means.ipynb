# Mean of Ratio vs Ratio of Means

Just a quick demonstration of the difference between the mean of ratios and the ratio of means.

The interesting case for insurance losses is when two consecutive age development losses have independent total residuals.

We use the log-normal distribution in this simulation, which is a common choice for insurance losses as it is strictly positive.

The below code defines a function which samples from the log-normal distribution.

```python
import numpy as np

np.random.seed(42)

def generate_sample_data():
    return np.exp(np.random.randn(100000) * 0.3 + 7.8)
```

Let's generate sample data for two consecutive age developments.

```python
a1 = generate_sample_data()
a2 = generate_sample_data()

print("Mean of a1:", np.mean(a1))
print("Mean of a2:", np.mean(a2))
print("Standard Deviation of a1:", np.std(a1))
print("Standard Deviation of a2:", np.std(a2))
```

Now let's calculate the mean ratio and the ratio of means.

```python
print("Ratio of medians of a1 and a2:", np.median(a2) / np.median(a1))
print("Ratio of means of a1 and a2:  ", np.mean(a2) / np.mean(a1))
print("Median of the ratio a2/a1:    ", np.median(a2 / a1))
print("Mean of the ratio a2/a1:      ", np.mean(a2 / a1))
```
